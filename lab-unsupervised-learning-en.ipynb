# Challenge 2 - Data Cleaning and Transformation

If your conclusion from the previous challenge is the data need cleaning/transformation, do it in the cells below. However, if your conclusion is the data need not be cleaned or transformed, feel free to skip this challenge. But if you do choose the latter, please provide rationale.

```python
from sklearn.preprocessing import StandardScaler

# --- Data Cleaning and Transformation ---

# 1. Handle Outliers using Log Transformation
numerical_columns = ['Fresh', 'Milk', 'Grocery', 'Frozen', 'Detergents_Paper', 'Delicassen']
df_log_transformed = df.copy()
df_log_transformed[numerical_columns] = np.log1p(df[numerical_columns])

# 2. Check for Highly Correlated Features
correlation_matrix = df_log_transformed.corr()
high_corr_features = correlation_matrix[(correlation_matrix > 0.9) & (correlation_matrix != 1.0)].stack()
print("Highly correlated features:", high_corr_features)

# Drop 'Detergents_Paper' if necessary
df_log_transformed.drop(columns=['Detergents_Paper'], inplace=True)
numerical_columns = [col for col in numerical_columns if col in df_log_transformed.columns]  # Update numerical columns list

# 3. Standardize the Data
scaler = StandardScaler()
df_scaled = df_log_transformed.copy()
df_scaled[numerical_columns] = scaler.fit_transform(df_log_transformed[numerical_columns])

# 4. Validate Transformation
print("\nTransformed Data Sample:\n", df_scaled.head())

# Visualize transformed distributions
for col in numerical_columns:
    plt.figure(figsize=(8, 4))
    sns.histplot(df_scaled[col], kde=True, bins=20)
    plt.title(f"Distribution after Scaling: {col}")
    plt.show()

# Check for Pareto Principle again
df_scaled['Total_Spending'] = df_scaled[numerical_columns].sum(axis=1)
df_sorted = df_scaled.sort_values('Total_Spending', ascending=False)
df_sorted['Cumulative_Percent'] = df_sorted['Total_Spending'].cumsum() / df_sorted['Total_Spending'].sum() * 100

# Plot Pareto principle
plt.figure(figsize=(10, 6))
plt.plot(df_sorted['Cumulative_Percent'], label='Cumulative Percentage')
plt.axhline(80, color='r', linestyle='--', label='80% Threshold')
plt.xlabel('Number of Customers (Sorted by Spending)')
plt.ylabel('Cumulative Percentage of Spending')
plt.title('Pareto Principle Analysis (After Transformation)')
plt.legend()
plt.show()

# Print customers who contribute to 80% of total spending
pareto_customers = df_sorted[df_sorted['Cumulative_Percent'] <= 80]
print("\nNumber of customers contributing to 80% of spending:", pareto_customers.shape[0])
print(pareto_customers)
```

# Bonus Challenge 2 - Changing K-Means Number of Clusters

As we mentioned earlier, we don't need to worry about the number of clusters with DBSCAN because it automatically decides that based on the parameters we send to it. But with K-Means, we have to supply the `n_clusters` param (if you don't supply `n_clusters`, the algorithm will use `8` by default). You need to know that the optimal number of clusters differs case by case based on the dataset. K-Means can perform badly if the wrong number of clusters is used.

In advanced machine learning, data scientists try different numbers of clusters and evaluate the results with statistical measures (read [here](https://en.wikipedia.org/wiki/Cluster_analysis#External_evaluation)). We are not using statistical measures today but we'll use our eyes instead. In the cells below, experiment with different number of clusters and visualize with scatter plots. What number of clusters seems to work best for K-Means?

```python
# Experiment with different numbers of clusters for K-Means
for n_clusters in range(2, 6):  # Trying 2 to 5 clusters
    kmeans = KMeans(n_clusters=n_clusters, random_state=42)
    customers_scale[f'labels_{n_clusters}'] = kmeans.fit_predict(customers_scale.drop(columns=['labels', 'labels_DBSCAN'], errors='ignore'))
    
    # Plot scatter for Detergents_Paper vs Milk
    plot_comparison('Detergents_Paper', 'Milk', f'labels_{n_clusters}', f'K-Means with {n_clusters} Clusters: Detergents Paper vs Milk')
```
